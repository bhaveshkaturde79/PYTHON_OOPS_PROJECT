**Employee Class**

```python
class Employee:
    def __init__(self, emp_id, name, age, department, salary):
        self.emp_id = emp_id
        self.name = name
        self.age = age
        self.department = department
        self.salary = salary
        self.attendance = []
        
    # calculate salary
    def calculate_salary(self, working_hours=160,
                         overtime_hours=0, bonus=0):
        hourly_rate = self.salary / 160

        overtime_pay = overtime_hours * hourly_rate * 1.5

        total_salary = (
            self.salary + overtime_pay + bonus
        )

        return round(total_salary, 2)

    # Mark_Attendence 
    def mark_attendance(self, date, status):
        status = status.capitalize()

        if status not in ["Present", "Absent", "Leave"]:
            print("Invalid attendance status!")
            return

        for record in self.attendance:
            if record["date"] == date:
                record["status"] = status
                print("Attendance updated successfully!")
                return

        self.attendance.append({
            "date": date,
            "status": status
        })

        print("Attendance marked successfully!")
        
    # Update_details
    def update_details(self, name=None, age=None,
                       department=None, salary=None):
        if name is not None:
            self.name = name

        if age is not None:
            self.age = age

        if department is not None:
            self.department = department

        if salary is not None:
            self.salary = salary
            
    # Display Information
    def display_info(self):
        print("\n--- Employee Details ---")
        print("Employee ID:", self.emp_id)
        print("Name:", self.name)
        print("Age:", self.age)
        print("Department:", self.department)
        print("Basic Salary:", self.salary)
        print("Calculated Salary:", self.calculate_salary())
        print("Attendance Records:", len(self.attendance))

emp1 = Employee("104", "Rupesh", 25, "CS", 30000)
emp1.mark_attendance("2026-10-15", "Present")

# Call Employee methods
emp1.display_info()

```

**Department Class**

```python
class Department:
    def __init__(self, name):
        self.name = name
        self.employees = []

    # Add Employee
    def add_employee(self, employee):
        if any(e.emp_id == employee.emp_id
               for e in self.employees):
            print("Employee already exists in department!")
            return

        self.employees.append(employee)
        employee.department = self.name
        print("Employee added to department successfully!")

    # Remove Employee
    def remove_employee(self, emp_id):
        for employee in self.employees:
            if employee.emp_id == emp_id:
                self.employees.remove(employee)
                print("Employee removed from department!")
                return True

        print("Employee not found in department!")
        return False

    # Display Department
    def display_department_info(self):
        print("\n--- Department Information ---")
        print("Department Name:", self.name)
        print("Total Employees:", len(self.employees))

    # List Employee
    def list_employees(self):
        print(f"\nEmployees in {self.name} Department:")

        if not self.employees:
            print("No employees in this department.")
            return

        for employee in self.employees:
            print(
                employee.emp_id,
                "-",
                employee.name
            )
dep1 = Department("CS")

# Add the existing employee object
dep1.add_employee(emp1)

# Test department methods
dep1.display_department_info()
dep1.list_employees()
```

**Company Class**

```python
class Company:
    def __init__(self, name):
        self.name = name
        self.employees = []
        self.departments = []

    # Add Department
    def add_department(self, department_name):
        if any(d.name.lower() == department_name.lower()
               for d in self.departments):
            print("Department already exists!")
            return

        department = Department(department_name)
        self.departments.append(department)
        print("Department created successfully!")

    # Remvove Department
    def remove_department(self, department_name):
        department = self.find_department(department_name)

        if department is None:
            print("Department not found!")
            return

        if department.employees:
            print("Cannot remove a department with employees.")
            return

        self.departments.remove(department)
        print("Department removed successfully!")

    # Find DEpartment
    def find_department(self, department_name):
        for department in self.departments:
            if department.name.lower() == department_name.lower():
                return department

        return None

    # Add Employee
    def add_employee(self, employee):
        if self.find_employee_by_id(employee.emp_id):
            print("Employee ID already exists!")
            return

        department = self.find_department(employee.department)

        if department is None:
            print("Department does not exist!")
            return

        self.employees.append(employee)
        department.add_employee(employee)
        print("Employee added to company successfully!")

    # Remove Employee
    def remove_employee(self, emp_id):
        employee = self.find_employee_by_id(emp_id)

        if employee is None:
            print("Employee not found!")
            return

        department = self.find_department(employee.department)

        if department:
            department.remove_employee(emp_id)

        self.employees.remove(employee)
        print("Employee removed from company successfully!")

    # Find Employee
    def find_employee_by_id(self, emp_id):
        for employee in self.employees:
            if employee.emp_id == emp_id:
                return employee

        return None

    # Update Employeee
    def update_employee_details(self, emp_id):
        employee = self.find_employee_by_id(emp_id)

        if employee is None:
            print("Employee not found!")
            return

        print("Leave a field blank to keep its current value.")

        name = input("New name: ").strip()
        age_input = input("New age: ").strip()
        department_name = input("New department: ").strip()
        salary_input = input("New basic salary: ").strip()

        try:
            age = int(age_input) if age_input else None
            salary = float(salary_input) if salary_input else None

            if age is not None and age <= 0:
                print("Age must be positive.")
                return

            if salary is not None and salary < 0:
                print("Salary cannot be negative.")
                return

        except ValueError:
            print("Age must be an integer and salary must be numeric.")
            return

        new_department = None

        if department_name:
            new_department = self.find_department(department_name)

            if new_department is None:
                print("Department does not exist!")
                return

        old_department = self.find_department(employee.department)

        if new_department and new_department != old_department:
            if old_department:
                old_department.remove_employee(emp_id)

            new_department.add_employee(employee)

        employee.update_details(
            name=name or None,
            age=age,
            department=(
                new_department.name
                if new_department
                else None
            ),
            salary=salary
        )

        print("Employee details updated successfully!")

    # Generate Reports
    def generate_reports(self):
        print("\n========== COMPANY REPORT ==========")
        print("Company:", self.name)
        print("Total Employees:", len(self.employees))

        total_salary = sum(
            employee.calculate_salary()
            for employee in self.employees
        )

        print("Total Basic Payroll:", round(total_salary, 2))

        print("\n--- Department-wise Distribution ---")

        for department in self.departments:
            print(
                department.name,
                ":",
                len(department.employees),
                "employees"
            )

        print("\n--- Employee Salary Report ---")

        for employee in self.employees:
            print(
                employee.name,
                "-",
                employee.calculate_salary()
            )

        print("\n--- Attendance Report ---")

        for employee in self.employees:
            present = sum(
                1 for record in employee.attendance
                if record["status"] == "Present"
            )

            absent = sum(
                1 for record in employee.attendance
                if record["status"] == "Absent"
            )

            leave = sum(
                1 for record in employee.attendance
                if record["status"] == "Leave"
            )

            print(
                employee.name,
                "| Present:", present,
                "| Absent:", absent,
                "| Leave:", leave
            )

c = Company("abcd")

c.add_department('CS')

c.add_employee(emp1)

found = c.find_employee_by_id("104")

if found:
    found.display_info()
else:
    print("Employee not found!")

c.generate_reports()
```

```python
def main():
    company = Company("ABC Company")
    company.add_department("CS")
    company.add_employee(emp1)
    while True:
        print("\n========== EMPLOYEE MANAGEMENT SYSTEM ==========")
        print("1. Add Department")
        print("2. Remove Department")
        print("3. Add Employee")
        print("4. Remove Employee")
        print("5. Search Employee")
        print("6. Update Employee Details")
        print("7. Mark Attendance")
        print("8. Calculate Employee Salary")
        print("9. List Department Employees")
        print("10. Display All Employees")
        print("11. Generate Reports")
        print("0. Exit")

        choice = input("Enter your choice: ").strip()

        try:
            if choice == "1":
                name = input("Enter department name: ").strip()

                if name:
                    company.add_department(name)
                else:
                    print("Department name cannot be empty.")

            elif choice == "2":
                name = input("Enter department name: ").strip()
                company.remove_department(name)

            elif choice == "3":
                emp_id = input("Enter employee ID: ").strip()

                if not emp_id:
                    print("Employee ID cannot be empty.")
                    continue

                if company.find_employee_by_id(emp_id):
                    print("Employee ID already exists!")
                    continue

                name = input("Enter employee name: ").strip()

                try:
                    age = int(input("Enter employee age: "))
                    salary = float(input("Enter basic salary: "))

                    if age <= 0 or salary < 0:
                        print("Age must be positive and salary non-negative.")
                        continue

                except ValueError:
                    print("Enter a valid age and salary.")
                    continue

                department = input("Enter department name: ").strip()

                if not name:
                    print("Employee name cannot be empty.")
                    continue

                employee = Employee(
                    emp_id, name, age, department, salary
                )

                company.add_employee(employee)

            elif choice == "4":
                emp_id = input("Enter employee ID: ").strip()
                company.remove_employee(emp_id)

            elif choice == "5":
                emp_id = input("Enter employee ID: ").strip()
                employee = company.find_employee_by_id(emp_id)

                if employee:
                    employee.display_info()
                else:
                    print("Employee not found!")

            elif choice == "6":
                emp_id = input("Enter employee ID: ").strip()
                company.update_employee_details(emp_id)

            elif choice == "7":
                emp_id = input("Enter employee ID: ").strip()
                employee = company.find_employee_by_id(emp_id)

                if employee:
                    date = input("Enter date (YYYY-MM-DD): ").strip()
                    status = input(
                        "Enter status (Present/Absent/Leave): "
                    ).strip()

                    employee.mark_attendance(date, status)
                else:
                    print("Employee not found!")

            elif choice == "8":
                emp_id = input("Enter employee ID: ").strip()
                employee = company.find_employee_by_id(emp_id)

                if employee:
                    try:
                        overtime = float(
                            input("Enter overtime hours: ") or 0
                        )
                        bonus = float(
                            input("Enter bonus amount: ") or 0
                        )

                        if overtime < 0 or bonus < 0:
                            print("Values cannot be negative.")
                            continue

                    except ValueError:
                        print("Enter valid numeric values.")
                        continue

                    salary = employee.calculate_salary(
                        overtime_hours=overtime,
                        bonus=bonus
                    )

                    print("Calculated Salary:", round(salary, 2))
                else:
                    print("Employee not found!")

            elif choice == "9":
                name = input("Enter department name: ").strip()
                department = company.find_department(name)

                if department:
                    department.display_department_info()
                    department.list_employees()
                else:
                    print("Department not found!")

            elif choice == "10":
                if not company.employees:
                    print("No employees found.")
                else:
                    for employee in company.employees:
                        employee.display_info()

            elif choice == "11":
                company.generate_reports()

            elif choice == "0":
                print("Thank you for using the system!")
                break

            else:
                print("Invalid choice! Please try again.")

        except (EOFError, KeyboardInterrupt):
            print("\nExiting Employee Management System.")
            break


if __name__ == "__main__":
    main()
   
```
